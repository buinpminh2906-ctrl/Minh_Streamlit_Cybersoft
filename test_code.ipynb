```python
import pandas as pd
import matplotlib.pyplot as plt



# Validate the relevant columns
cols = ["Product Category", "Total Amount"]
print(df[cols].isnull().sum())
print(df[cols].dtypes)

# Aggregate sales and keep up to the top 10 categories
sales_by_category = (
    df.dropna(subset=cols)
      .groupby("Product Category")["Total Amount"]
      .sum()
      .nlargest(10)
)

plt.figure(figsize=(10, 6))
plt.pie(
    sales_by_category,
    labels=sales_by_category.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Share of Total Sales by Product Category")
plt.tight_layout()
plt.show()# Validate the relevant columns
cols = ["Product Category", "Total Amount"]
print(df[cols].isnull().sum())
print(df[cols].dtypes)

# Aggregate sales and keep up to the top 10 categories
sales_by_category = (
    df.dropna(subset=cols)
      .groupby("Product Category")["Total Amount"]
      .sum()
      .nlargest(10)
)

plt.figure(figsize=(10, 6))
plt.pie(
    sales_by_category,
    labels=sales_by_category.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Share of Total Sales by Product Category")
plt.tight_layout()
plt.show()

```
